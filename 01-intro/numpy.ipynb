### Creating Arrays

```python
np.zeros(10)
```

```python
np.ones(10)
```

```python
np.full(10, 5) # Array of 10 elements, all set to 5
```

```python
a = np.array([1, 2, 3, 4, 5])
a
```

```python
a[2] = 10
a
```

```python
np.arange(3, 10) # Array of integers from 3 to 9
```

```python
np.linspace(0, 1, 5) # Array of 5 evenly spaced numbers between 0 and 1
```

```python
# Converting list to numpy array
list1 = [1, 2, 3, 4, 5]
array1 = np.array(list1)
array1
```

### Multidimensional Arrays

```python
np.zeros((3, 2)) # 10 rows, 2 columns
```

```python
n = np.array([[1, 2, 3], [4, 5, 6]])
n
```

```python
n[0, 1] = 100
n
```

```python
n[1] = [7, 8, 9]
n
```

```python
n[:, 1] = 99 # All rows, second column 
n
```

### Generating Random Arrays

```python
np.random.seed(2) # Set the random seed for reproducibility
random_array = np.random.rand(3, 2) # 3 rows, 2 columns
random_array
```

```python
normal_distribution_array = np.random.randn(3, 2) # 3 rows, 2 columns
normal_distribution_array
```

```python
random_integers_array = np.random.randint(low=0, high=10, size=(3, 2)) # 3 rows, 2 columns, random integers between 0 and 9
random_integers_array
```

### Array Operations

#### Element-wise Operations

```python
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr1
```

```python
arr2 = arr1 + 10
arr2
```

```python
arr2 / arr1
```

#### Comparison Operations

```python
arr1
```

```python
arr1 > 3
```

```python
arr2 = np.array([[0, 2, 2], [10, 4, 12]])
arr2
```

```python
arr1 > arr2
```

```python
arr1[arr1 > arr2]
```

#### Summary Operations

```python
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr1
```

```python
min_value = np.min(arr1)
max_value = np.max(arr1)
sum_value = np.sum(arr1)
mean_value = np.mean(arr1)
std_dev_value = np.std(arr1)

print("Minimum value:", min_value)
print("Maximum value:", max_value)
print("Sum of all elements:", sum_value)
print("Mean value:", mean_value)
print("Standard deviation:", std_dev_value)
```
