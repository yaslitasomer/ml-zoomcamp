#### Multiplication

```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])
u.shape
```

```python
def vector_vector_multiplication(u, v):
    assert u.shape == v.shape, "Vectors must have the same shape"
    
    n = u.shape[0]  # Number of elements in the vectors
    result = 0
    
    for i in range(n):
        result += u[i] * v[i]
        
    return result
```

```python
vector_vector_multiplication(u, v)
```

```python
def matrix_vector_multiplication(A, u):
    assert A.shape[1] == u.shape[0], "Incompatible shapes for matrix-vector multiplication"
    
    m = A.shape[0]  # Number of rows in the matrix
    n = A.shape[1]  # Number of columns in the matrix
    result = np.zeros(m)
    
    for i in range(m):
        for j in range(n):
            result[i] += A[i, j] * u[j]
            
    return result
```

```python
U = np.array([[1, 2, 3], [4, 5, 6]])
matrix_vector_multiplication(U, u)
```

```python
# Dot product implementation
def matrix_matrix_multiplication(A, B):
    assert A.shape[1] == B.shape[0], "Incompatible shapes for matrix-matrix multiplication"
    
    num_rows = A.shape[0]
    num_cols = B.shape[1]
    
    results = np.zeros((num_rows, num_cols))
    
    for i in range(num_cols):
        bi = B[:, i]  # Get the i-th column of B
        Abi = matrix_vector_multiplication(A, bi)  # Multiply A with the i-th column of B
        results[:, i] = Abi  # Store the result in the i-th column of the results matrix
        
    return results
```

```python
B = np.array([[7, 8], [9, 10], [11, 12]])
matrix_matrix_multiplication(U, B)
```

```python
# Using NumPy's built-in functions for matrix multiplication
np.dot(U, B)
```

#### Inverse

```python
U = np.array([[1, 2, 3], [4, 5, 6], [7, 8, 11]])
U
```

```python
U_inv = np.linalg.inv(U)
U_inv   
```

```python
U.dot(U_inv)  # This should return the identity matrix
```
